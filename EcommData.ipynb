```python
import pandas as pd
```

```python
df = pd.read_csv("EcommData_CSV.csv", sep=";")
```

```python
pd.set_option('display.max_colwidth', None)
pd.set_option('display.max_rows', None)
```

```python
df.info()
```

```python
df.head()
#Purchase Amount (USD) and Review Rating uses a comma as the decimal separator, so we need to convert it to a float. We can do this by replacing the comma with a dot and then converting the column to a float.
#Purchase Date is in the format of dd.mm.yyyy, so we need to convert it to a datetime format. We can do this by using the pd.to_datetime() function and specifying the format.
```

```python
df["Purchase Amount (USD)"] = df["Purchase Amount (USD)"].str.replace(",", ".").astype(float)
df["Review Rating"] = df["Review Rating"].str.replace(",", ".").astype(float)
df["Purchase Date"] = pd.to_datetime(df["Purchase Date"], format="%d.%m.%Y")
```

```python
df.head()
```

```python
df.isnull().sum()
#There are no missing values, so we can continue with the analysis.
```

```python
#making all the column names lowercase and replacing spaces with underscores for better accessibility
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df = df.rename(columns={'purchase_amount_(usd)': 'purchase_amount_usd'})
```

```python
df.columns
```

```python
#create a column age_group to categorize customers into different age groups based on their age. The age groups are defined as follows:
bins = [17, 29, 44, 59, 70]
labels = ['Young Adult', 'Adult', 'Middle Aged', 'Senior']
df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels, include_lowest=True)
```

```python
##Splitting the date purchased column into one for year, and month to be able to analyze the data by year and month
df['purchase_year'] = df['purchase_date'].dt.year
df['purchase_month'] = df['purchase_date'].dt.month
##Also adding the quarter of the year to be able to analyze the data by quarter
# Extract the quarter (1, 2, 3, or 4)
df['quarter'] = df['purchase_date'].dt.quarter
# Create a combined year-quarter column
df['year_quarter'] = (
    df['purchase_date'].dt.year.astype(str)
    + '-Q'
    + df['purchase_date'].dt.quarter.astype(str)
)
```

```python
print(df[['purchase_date', 'purchase_year', 'quarter', 'year_quarter']].head(10))
```

```python
#checking for duplicates in the data. We can do this by using the duplicated() function and then summing the result to get the total number of duplicates.
df.duplicated().sum()
```

```python
#checking if the customer_id column has unique values. We can do this by using the nunique() function and comparing it to the total number of rows in the dataframe. If the two values are equal, then the customer_id column has unique values.
df['customer_id'].nunique() == len(df)
```

```python
from sqlalchemy import create_engine

#Connecting to PostgreSQL database using SQLAlchemy. We will use the create_engine() function to create a connection to the database. 
username = "postgres" #default username for PostgreSQL
password = "password" #replace with your password
host = "localhost" #replace with your host
port = "5432" #default port for PostgreSQL
database = "EcommData" #replace with your database name

engine = create_engine(f'postgresql+psycopg2://{username}:{password}@{host}:{port}/{database}')

#Load DataFrame into PostgreSQL database using the to_sql() function. We will specify the name of the table, the connection to the database, and the if_exists parameter to specify what to do if the table already exists. We will set if_exists to 'replace' to replace the table if it already exists.
table_name = "consumer"
df.to_sql(table_name, engine, if_exists='replace', index=False)

print(f"DataFrame loaded into table '{table_name}' in PostgreSQL database '{database}'.")
```

```python
# Exporting the SQL table to a CSV file to be able to use it in PowerBI for visualization. 
# We will use the to_csv() function to export the table to a CSV file. We will specify the name of the CSV file and set index to False to not include the index in the CSV file.
# Reading PostgreSQL table
df = pd.read_sql_query("SELECT * FROM consumer", engine)
df.to_csv("consumer_data.csv", index=False)

print("SQL table exported to CSV file 'consumer_data.csv'.")
```
